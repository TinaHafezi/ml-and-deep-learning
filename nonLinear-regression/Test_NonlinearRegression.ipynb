```python
import numpy as np 
import pandas as pd 
```

```python
#خواندن دیتاست Test
test_df=pd.read_csv("auto-test-preprocessed.csv")

print("Shape:",test_df.shape)
print(test_df.head())
```

```python
#جدا کردن ورودی و خروجی 
X_test = test_df["HorsePower"].values.reshape(-1,1)
y_test = test_df["MPG"].values.reshape(-1,1)
print("X shape:",X_test.shape)
print("y shape:",y_test.shape)
print(f"[Test] تعداد نمونه ها:{X_test.shape[0]}")
```

```python
#لودکردن مدل 
w=np.load("./Best_Model.npy")
print("Trained weights:")
print(w)

print("weight shape:",w.shape)
```

```python
#مشخص کردن درجه مدل 
degree=w.shape[0] - 1
print("Degree:",degree)
```

```python

def polynomial_feature_mapping(X,degree):
    n=X.shape[0]
    Phi=np.ones((n,degree + 1))

    for j in range(1,degree + 1):
        Phi[:,j]=X[:,0]**j

    return Phi

```

```python
degree=3
Phi_test=polynomial_feature_mapping(X_test ,degree)
print("Phi_test shape:",Phi_test.shape)
```

```python
# پیش بینی

y_hat = Phi_test @ w

print("Predictions:")
print(y_hat)
```

```python
#محاسبه خطا 
M=np.sum(np.abs(y_hat - y_test))
print("Test Error:",M)
```
